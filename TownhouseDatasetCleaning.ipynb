```python
#import statements
import pandas as pd
```

```python
#reading in the original townhouse dataset
originalTownhouse_df = pd.read_csv("TownHouse_Redfin.csv")
```

```python
originalTownhouse_df.head()
```

```python
#townhouse dataset shape: 10,710 rows x 28 columns
originalTownhouse_df.shape
```

```python
#double checking that property type column is present in the townhouse dataset
if 'PROPERTY TYPE' in originalTownhouse_df.columns:
    print("property type column is in townhouse dataset")
else:
    print("no property type column in townhouse dataset")
```

```python
#checking if na values are getting recognized

#Check that NA values are recognized as missing values:
print("Number of Recognized Missing Values (NA) : ", originalTownhouse_df.isna().sum())
```

```python
#creating the cleaned townhouse dataframe copy - modify this dataset moving forward in next steps
cleanedTownhouse_df = originalTownhouse_df.copy()
```

```python
cleanedTownhouse_df.head()
```

```python
#dropping 'LAST UPDATED' and 'REGION TYPE' columns from cleaned townhouse dataset:
cleanedTownhouse_df = cleanedTownhouse_df.drop(columns = ['LAST UPDATED', 'REGION TYPE'])
```

```python
cleanedTownhouse_df.head()
```

```python
#VERIFICATION: double checking that 'LAST UPDATED' and 'REGION TYPE' columns were dropped from cleaned dataset successfully

dropped_cols = {'LAST UPDATED', 'REGION TYPE'}

if dropped_cols.issubset(cleanedTownhouse_df.columns):
    print("dropped columns are present!")
else:
    print("dropped columns are not present!")
```

```python
#converting period begin and period end to pandas datetime objects
cleanedTownhouse_df['PERIOD BEGIN'] = pd.to_datetime(cleanedTownhouse_df['PERIOD BEGIN'])
cleanedTownhouse_df['PERIOD END'] = pd.to_datetime(cleanedTownhouse_df['PERIOD END'])
```

```python
#Verifying that period begin and period end were properly date time formatted:
print(cleanedTownhouse_df['PERIOD BEGIN'].dtype)
print(cleanedTownhouse_df['PERIOD END'].dtype)
```

```python
cleanedTownhouse_df['Year'] = cleanedTownhouse_df['PERIOD BEGIN'].dt.year
cleanedTownhouse_df['Month'] = cleanedTownhouse_df['PERIOD BEGIN'].dt.month
cleanedTownhouse_df['Month_Name'] = cleanedTownhouse_df['PERIOD BEGIN'].dt.month_name()
cleanedTownhouse_df['Quarter'] = cleanedTownhouse_df['PERIOD BEGIN'].dt.quarter
```

```python
#VERIFICATION: double checking that added date year, month, month name, and quarter columns are present in townhouse dataset

addedDate_cols = {'Year', 'Month', 'Month_Name', 'Quarter'}

if addedDate_cols.issubset(cleanedTownhouse_df.columns):
    print("added date columns are present!")
else:
    print("added date columns are not present!")
```

```python
percent_cols = [
    'DOWN PAYMENT AMOUNT ($)',
    'INCOME REQUIRED YOY (%)',
    'SHARE OF INCOME ON HOUSING (%)',
    'SHARE OF INCOME ON HOUSING YOY (PPTS)',
    'AFFORDABLE LISTING SHARE (%)',
    'AFFORDABLE LISTING SHARE YOY (PPTS)',
    'YEARS TO SAVE DOWN PAYMENT YOY (%)',
    'MEDIAN SALE PRICE YOY (%)',
    'HOMES SOLD YOY (%)'
]
```

```python
#double checking that the data types of percent_cols are currently float values:
for col in percent_cols:
    print("col name: ", col, " data type: ", cleanedTownhouse_df[col].dtype)
```

```python
#proportionalizing the percent columns, giving them a new column (a copy of the original percentage col / 100)
for col in percent_cols:
    newColName = col + ' Prop'
    cleanedTownhouse_df[newColName] = cleanedTownhouse_df[col] / 100
```

```python
#verifying that those proportion columns were made, and check their data types are float values

cleanedTownhouse_df.columns
```

```python
print(cleanedTownhouse_df.dtypes)
```

```python
#converting median household income to float
cleanedTownhouse_df['MEDIAN HOUSEHOLD INCOME ($)'] = cleanedTownhouse_df['MEDIAN HOUSEHOLD INCOME ($)'].astype(float)
```

```python
cleanedTownhouse_df['MEDIAN HOUSEHOLD INCOME ($)'].dtype
```

```python
#seasonal column mapping:

season_map = {
    12: 'Winter', 1: 'Winter', 2: 'Winter',
    3: 'Spring', 4: 'Spring', 5: 'Spring',
    6: 'Summer', 7: 'Summer', 8: 'Summer',
    9: 'Fall', 10: 'Fall', 11: 'Fall'
}

#Creating a season column
cleanedTownhouse_df['Season'] = cleanedTownhouse_df['Month'].map(season_map)
```

```python
season_col = {'Season'}

if season_col.issubset(cleanedTownhouse_df.columns):
    print("added season column is present!")
else:
    print("added season column is not present!")
```
