```python
#import statements
import pandas as pd
```

```python
originalSingleFamily_df = pd.read_csv("SingleFamily_Redfin.csv")
```

```python
#inspecting the head of the original single family dataframe
originalSingleFamily_df.head()
```

```python
#shape of original dataset: 10,921 rows x 28
originalSingleFamily_df.shape
```

```python
#double checking that property type column is present in the dataset
if 'PROPERTY TYPE' in originalSingleFamily_df.columns:
    print("property type column is in single family dataset")
else:
    print("no property type column in single family dataset")
```

```python
#Check that NA values are recognized as missing values:
print("Number of Recognized Missing Values (NA) : ", originalSingleFamily_df.isna().sum())
```

```python
#creating the cleaned single family dataframe copy - modify this dataset moving forward in next steps
cleanedSingleFamily_df = originalSingleFamily_df.copy()
```

```python
#double-checking head of dataset
cleanedSingleFamily_df.head()
```

```python
#dropping 'LAST UPDATED' and 'REGION TYPE' columns from cleaned dataset:
cleanedSingleFamily_df = cleanedSingleFamily_df.drop(columns = ['LAST UPDATED', 'REGION TYPE'])
```

```python
cleanedSingleFamily_df.head()
```

```python
#VERIFICATION: double checking that 'LAST UPDATED' and 'REGION TYPE' columns were dropped from cleaned dataset successfully

dropped_cols = {'LAST UPDATED', 'REGION TYPE'}

if dropped_cols.issubset(cleanedSingleFamily_df.columns):
    print("dropped columns are present!")
else:
    print("dropped columns are not present!")
```

```python
#Pandas Date Time Formatting:

#Converting Period Begin and Period End to pandas date time objects
cleanedSingleFamily_df['PERIOD BEGIN'] = pd.to_datetime(cleanedSingleFamily_df['PERIOD BEGIN'])
cleanedSingleFamily_df['PERIOD END'] = pd.to_datetime(cleanedSingleFamily_df['PERIOD END'])
```

```python
#Verifying that period begin and period end were properly date time formatted:
print(cleanedSingleFamily_df['PERIOD BEGIN'].dtype)
print(cleanedSingleFamily_df['PERIOD END'].dtype)
```

```python
#creating new columns from pandas datetime columns for seasonality
cleanedSingleFamily_df['Year'] = cleanedSingleFamily_df['PERIOD BEGIN'].dt.year
cleanedSingleFamily_df['Month'] = cleanedSingleFamily_df['PERIOD BEGIN'].dt.month
cleanedSingleFamily_df['Month_Name'] = cleanedSingleFamily_df['PERIOD BEGIN'].dt.month_name()
cleanedSingleFamily_df['Quarter'] = cleanedSingleFamily_df['PERIOD BEGIN'].dt.quarter
```

```python
#VERIFICATION: double checking that added date year, month, month name, and quarter columns are present

addedDate_cols = {'Year', 'Month', 'Month_Name', 'Quarter'}

if addedDate_cols.issubset(cleanedSingleFamily_df.columns):
    print("added date columns are present!")
else:
    print("added date columns are not present!")
```

```python
cleanedSingleFamily_df.columns
```

```python
percent_cols = [
    'DOWN PAYMENT AMOUNT ($)',
    'INCOME REQUIRED YOY (%)',
    'SHARE OF INCOME ON HOUSING (%)',
    'SHARE OF INCOME ON HOUSING YOY (PPTS)',
    'AFFORDABLE LISTING SHARE (%)',
    'AFFORDABLE LISTING SHARE YOY (PPTS)',
    'YEARS TO SAVE DOWN PAYMENT YOY (%)',
    'MEDIAN SALE PRICE YOY (%)',
    'HOMES SOLD YOY (%)'
]
```

```python
#double checking that the data types of percent_cols are currently float values:
for col in percent_cols:
    print("col name: ", col, " data type: ", cleanedSingleFamily_df[col].dtype)
```

```python
#proportionalizing the percent columns, giving them a new column (a copy of the original percentage col / 100)
for col in percent_cols:
    newColName = col + ' Prop'
    cleanedSingleFamily_df[newColName] = cleanedSingleFamily_df[col] / 100
```

```python
#verifying that those proportion columns were made, and check their data types are float values

cleanedSingleFamily_df.columns
```

```python
print(cleanedSingleFamily_df.dtypes)
```

```python
#converting median household income to float
cleanedSingleFamily_df['MEDIAN HOUSEHOLD INCOME ($)'] = cleanedSingleFamily_df['MEDIAN HOUSEHOLD INCOME ($)'].astype(float)
```

```python
cleanedSingleFamily_df['MEDIAN HOUSEHOLD INCOME ($)'].dtype
```

```python
#seasonal column mapping:

season_map = {
    12: 'Winter', 1: 'Winter', 2: 'Winter',
    3: 'Spring', 4: 'Spring', 5: 'Spring',
    6: 'Summer', 7: 'Summer', 8: 'Summer',
    9: 'Fall', 10: 'Fall', 11: 'Fall'
}

#Creating a season column
cleanedSingleFamily_df['Season'] = cleanedSingleFamily_df['Month'].map(season_map)
```

```python
season_col = {'Season'}

if season_col.issubset(cleanedSingleFamily_df.columns):
    print("added season column is present!")
else:
    print("added season column is not present!")
```

```python
#finally seeing the cleanedSingleFamily_df
cleanedSingleFamily_df.head()
```

```python
#downloading the cleaned notebook
cleanedSingleFamily_df.to_csv('SingleFamily_cleanedDF.csv', index=False)
```
