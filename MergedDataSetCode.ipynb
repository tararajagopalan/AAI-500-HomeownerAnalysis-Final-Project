**Alterations Needed for Every Dataset**

* Justin: Condo and Starter Home

    * Add prop columns in


* Tara: Single Family and Townhouse

    * change Month_Name to Month Name (rename column)
    * Add Year-Month col
    * Add Period Length (Days)

```python
#the percent columns missing for justins dataset
percent_cols = [
    'DOWN PAYMENT AMOUNT ($)',
    'INCOME REQUIRED YOY (%)',
    'SHARE OF INCOME ON HOUSING (%)',
    'SHARE OF INCOME ON HOUSING YOY (PPTS)',
    'AFFORDABLE LISTING SHARE (%)',
    'AFFORDABLE LISTING SHARE YOY (PPTS)',
    'YEARS TO SAVE DOWN PAYMENT YOY (%)',
    'MEDIAN SALE PRICE YOY (%)',
    'HOMES SOLD YOY (%)'
]
```

```python
#double checking that the data types of percent_cols are currently float values:
for col in percent_cols:
    print("col name: ", col, " data type: ", Condo_df[col].dtype)
```

```python
#double checking that the data types of percent_cols are currently float values:
for col in percent_cols:
    print("col name: ", col, " data type: ", StarterHome_df[col].dtype)
```

```python
#Condo: proportionalizing the percent columns, giving them a new column (a copy of the original percentage col / 100)
for col in percent_cols:
    newColName = col + ' Prop'
    Condo_df[newColName] = Condo_df[col] / 100
```

```python
#Starter Home: proportionalizing the percent columns, giving them a new column (a copy of the original percentage col / 100)
for col in percent_cols:
    newColName = col + ' Prop'
    StarterHome_df[newColName] = StarterHome_df[col] / 100
```

```python
SingleFamily_df['PERIOD BEGIN'] = pd.to_datetime(SingleFamily_df['PERIOD BEGIN'])
Townhouse_df['PERIOD BEGIN'] = pd.to_datetime(Townhouse_df['PERIOD BEGIN'])
```

```python
SingleFamily_df['Year-Month'] = SingleFamily_df['PERIOD BEGIN'].dt.strftime('%Y-%m')
Townhouse_df['Year-Month'] = Townhouse_df['PERIOD BEGIN'].dt.strftime('%Y-%m')
```

```python
SingleFamily_df['PERIOD END'] = pd.to_datetime(SingleFamily_df['PERIOD END'])
Townhouse_df['PERIOD END'] = pd.to_datetime(Townhouse_df['PERIOD END'])
```

```python
#Creating period length in days variable for Tara's Datasets: Single Family and Townhouse
SingleFamily_df['Period Length (Days)'] = (SingleFamily_df['PERIOD END'] - SingleFamily_df['PERIOD BEGIN']).dt.days + 1
Townhouse_df['Period Length (Days)'] = (Townhouse_df['PERIOD END'] - Townhouse_df['PERIOD BEGIN']).dt.days + 1
```
